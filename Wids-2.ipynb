```python
from sklearn.utils import shuffle

# Step 2: Count current class distribution
current_counts = train_targets.groupby(['ADHD_Outcome', 'Sex_F']).size()
majority_count = current_counts.max()

# Step 3: Initialize list for augmented data
augmented_rows = []

# Step 4: For each class, if it's underrepresented, add rows from train_2fmri
for (adhd, sex), count in current_counts.items():
    if count < majority_count:
        needed = majority_count - count
        available_rows = train_2fmri_full[
            (train_2fmri_full['ADHD_Outcome'] == adhd) &
            (train_2fmri_full['Sex_F'] == sex)
        ]
        to_add = available_rows.sample(n=min(needed, len(available_rows)), random_state=42)
        augmented_rows.append(to_add)
```

```python
# Merge metadata with current and augmented rows
train_with_meta = train_fmri_full.merge(train_quantitative, on='participant_id')
augmented_df = pd.concat(augmented_rows, ignore_index=True) if augmented_rows else pd.DataFrame()
augmented_with_meta = augmented_df.merge(old_train_quantitative, on='participant_id')

# Combine both
final_df = pd.concat([train_with_meta, augmented_with_meta], ignore_index=True)
final_df = shuffle(final_df, random_state=42).reset_index(drop=True)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import mutual_info_classif, SelectKBest
from sklearn.calibration import CalibratedClassifierCV
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models, Input, regularizers, optimizers
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import f1_score
import pandas as pd

# Set global seed for reproducibility
SEED = 42
tf.random.set_seed(SEED)
np.random.seed(SEED)

# ========== DATA PREPROCESSING ==========
# Step 1: Separate fMRI and metadata
corr_cols = [col for col in final_df.columns if 'throw' in col]
meta_cols = ['participant_id', 'ADHD_Outcome', 'Sex_F']
X_corr = final_df[corr_cols].copy()
X_quant = final_df.drop(columns=corr_cols + meta_cols).copy()

# Step 2: Handle missing values
X_quant.fillna(X_quant.mean(), inplace=True)

# Step 3: Row-wise scaling of correlation matrices
X_corr_scaled = np.array([StandardScaler().fit_transform(row.reshape(-1, 1)).flatten() for row in X_corr.values])

# Scaling for other quantitative features
scaler_quant = StandardScaler()
X_quant_scaled = scaler_quant.fit_transform(X_quant)

# Step 4: Feature selection
# For fMRI data
mi_adhd = mutual_info_classif(X_corr_scaled, final_df['ADHD_Outcome'], random_state=SEED)
mi_sex = mutual_info_classif(X_corr_scaled, final_df['Sex_F'], random_state=SEED)
mi_combined = 0.5 * mi_adhd + 0.5 * mi_sex  # Weighted combination
top_k_corr_features = 1024
top_k_indices = np.argsort(mi_combined)[-top_k_corr_features:]
X_corr_selected = X_corr_scaled[:, top_k_indices]
X_corr_cnn = X_corr_selected.reshape(-1, 32, 32, 1)

selector_adhd = SelectKBest(mutual_info_classif, k=10)
selector_sex = SelectKBest(mutual_info_classif, k=15)  # More features for Sex_F
X_quant_adhd = selector_adhd.fit_transform(X_quant_scaled, final_df['ADHD_Outcome'])
X_quant_sex = selector_sex.fit_transform(X_quant_scaled, final_df['Sex_F'])

    # Combine indices with duplicate Sex_F features for emphasis
combined_indices = np.unique(np.concatenate((
        selector_adhd.get_support(indices=True),
        selector_sex.get_support(indices=True))))

X_quant_selected = X_quant_scaled[:, combined_indices]

# Labels
y_adhd = final_df['ADHD_Outcome'].values
y_sex = final_df['Sex_F'].values
y_strat = (y_adhd << 1) + y_sex  # For stratified splitting
```

```python
SEED = 42

# ========== MODEL ARCHITECTURE ==========
def create_enhanced_model(input_cnn_shape, input_quant_shape):
    # CNN Branch (fMRI) - ADHD focused
    input_cnn = Input(shape=input_cnn_shape, name='fmri_input')
    x = layers.Conv2D(32, (3, 3), activation='relu', padding='same')(input_cnn)
    x = layers.BatchNormalization()(x)
    x = layers.MaxPooling2D((2, 2))(x)
    x = layers.Dropout(0.3)(x)

    x = layers.Conv2D(64, (3, 3), activation='relu', padding='same')(x)
    x = layers.BatchNormalization()(x)
    x = layers.MaxPooling2D((2, 2))(x)
    x = layers.Dropout(0.4)(x)

    x = layers.Flatten()(x)
    x = layers.Dense(128, activation='relu', kernel_regularizer=regularizers.l2(0.001))(x)
    x = layers.BatchNormalization()(x)

    # Enhanced DNN Branch (metadata) - Sex focused
    input_quant = Input(shape=input_quant_shape, name='quant_input')
    y = layers.Dense(64, activation='relu')(input_quant)
    y = layers.BatchNormalization()(y)
    y = layers.Dropout(0.3)(y)
    y = layers.Dense(64, activation='relu')(y)
    y = layers.BatchNormalization()(y)

    # Separate branch for Sex prediction
    y_sex = layers.Dense(32, activation='relu')(y)
    y_sex = layers.BatchNormalization()(y_sex)

    # Merge branches
    combined = layers.concatenate([x, y])
    z = layers.Dense(64, activation='relu')(combined)
    z = layers.BatchNormalization()(z)
    z = layers.Dropout(0.3)(z)

    # Output layers
    output_adhd = layers.Dense(1, activation='sigmoid', name='ADHD_Outcome')(z)
    output_sex = layers.Dense(1, activation='sigmoid', name='Sex_F')(y_sex)  # Direct from sex branch

    model = models.Model(inputs=[input_cnn, input_quant], outputs=[output_adhd, output_sex])

    # Weighted loss (70% ADHD, 30% Sex)
    model.compile(
        optimizer=optimizers.Adam(learning_rate=0.001),
        loss={
            'ADHD_Outcome': 'binary_crossentropy',
            'Sex_F': 'binary_crossentropy'
        },
        metrics={
            'ADHD_Outcome': ['accuracy', tf.keras.metrics.AUC(name='auc')],
            'Sex_F': ['accuracy', tf.keras.metrics.AUC(name='auc')]
        },
        loss_weights={'ADHD_Outcome': 0.5, 'Sex_F': 0.5}
    )
    return model

```

```python
# ========== CROSS-VALIDATION ==========
thresholds = np.linspace(0.1, 0.9, 81)
f1_scores_adhd = []
f1_scores_sex = []
best_thresholds_adhd = []
best_thresholds_sex = []

kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)

for train_idx, val_idx in kf.split(X_corr_cnn, y_strat):
    X_train_cnn, X_val_cnn = X_corr_cnn[train_idx], X_corr_cnn[val_idx]
    X_train_quant, X_val_quant = X_quant_selected[train_idx], X_quant_selected[val_idx]
    y_train_adhd, y_val_adhd = y_adhd[train_idx], y_adhd[val_idx]
    y_train_sex, y_val_sex = y_sex[train_idx], y_sex[val_idx]

    # Build and train the model
    model = create_enhanced_model((32, 32, 1), X_train_quant.shape[1:])
    early_stop = tf.keras.callbacks.EarlyStopping(
        monitor='val_loss',
        patience=10,
        restore_best_weights=True,
        min_delta=0.001
    )

    # Learning rate scheduler
    lr_scheduler = tf.keras.callbacks.ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=3,
        min_lr=1e-6
    )

    model.fit(
        [X_train_cnn, X_train_quant],
        [y_train_adhd, y_train_sex],
        validation_data=([X_val_cnn, X_val_quant], [y_val_adhd, y_val_sex]),
        epochs=50,
        batch_size=32,
        callbacks=[early_stop, lr_scheduler],
        verbose=0
    )

    # Predict probabilities
    pred_adhd_probs, pred_sex_probs = model.predict([X_val_cnn, X_val_quant], verbose=0)

    # Find optimal thresholds
    best_f1_adhd, best_thresh_adhd = 0, 0.5
    best_f1_sex, best_thresh_sex = 0, 0.5

    for t in thresholds:
        f1_adhd = f1_score(y_val_adhd, (pred_adhd_probs > t).astype(int))
        f1_sex = f1_score(y_val_sex, (pred_sex_probs > t).astype(int))

        if f1_adhd > best_f1_adhd:
            best_f1_adhd = f1_adhd
            best_thresh_adhd = t

        if f1_sex > best_f1_sex:
            best_f1_sex = f1_sex
            best_thresh_sex = t

    f1_scores_adhd.append(best_f1_adhd)
    f1_scores_sex.append(best_f1_sex)
    best_thresholds_adhd.append(best_thresh_adhd)
    best_thresholds_sex.append(best_thresh_sex)

print(f"✅ Mean F1 ADHD: {np.mean(f1_scores_adhd):.4f} (±{np.std(f1_scores_adhd):.4f})")
print(f"✅ Mean F1 Sex: {np.mean(f1_scores_sex):.4f} (±{np.std(f1_scores_sex):.4f})")
```

```python

SEED = 42

# ========== FINAL MODEL TRAINING ==========
avg_thresh_adhd = np.mean(best_thresholds_adhd)
avg_thresh_sex = np.mean(best_thresholds_sex)

final_model = create_enhanced_model((32, 32, 1), X_quant_selected.shape[1:])
early_stop = tf.keras.callbacks.EarlyStopping(
    monitor='val_loss',
    patience=10,
    restore_best_weights=True
)

history = final_model.fit(
    [X_corr_cnn, X_quant_selected],
    [y_adhd, y_sex],
    validation_split=0.2,
    epochs=50,
    batch_size=32,
    callbacks=[early_stop],
    verbose=1
)
```

```python
SEED = 42

# ========== TEST DATA PROCESSING ==========
X_test_corr = test_df[corr_cols].copy()
X_test_corr_scaled = np.array([StandardScaler().fit_transform(row.reshape(-1, 1)).flatten() for row in X_test_corr.values])
X_test_corr_selected = X_test_corr_scaled[:, top_k_indices]
X_test_cnn = X_test_corr_selected.reshape(-1, 32, 32, 1)

X_test_quant = test_df.drop(columns=corr_cols + ['participant_id'], errors='ignore').copy()
X_test_quant_scaled = scaler_quant.transform(X_test_quant)
X_test_quant_selected = X_test_quant_scaled[:, combined_indices]

# ========== FINAL PREDICTION ==========
pred_probs = final_model.predict([X_test_cnn, X_test_quant_selected], verbose=0)
pred_classes = (
    (pred_probs[0] > avg_thresh_adhd).astype(int),
    (pred_probs[1] > avg_thresh_sex).astype(int)
)

# ========== SAVE RESULTS ==========
output_df = pd.DataFrame({
    'participant_id': test_df['participant_id'],
    'ADHD_Prob': pred_probs[0].flatten(),
    'ADHD_Outcome': pred_classes[0].flatten(),
    'Sex_Prob': pred_probs[1].flatten(),
    'Sex_F': pred_classes[1].flatten()
})

output_df.to_csv('Enhanced_Final_predictions.csv', index=False)
```
